```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt

# Load the dataset
digits = load_digits()

# Display dataset information
print("Dataset Shape:", digits.data.shape)
print("Image Shape:", digits.images.shape)
print("Target Labels:", digits.target_names)

# Display the first image
plt.imshow(digits.images[1], cmap='pink')
plt.title(f"Label: {digits.target[1]}")
plt.axis('off')
plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
#Load the dataset
digits = load_digits()
#Create a 2*5 grid of subplots
fig, axes=plt.subplots(2,5, figsize=(10,5), layout="constrained")
#Iterate through axes and display the first 10 digits
for i,ax in enumerate(axes.flat):
    ax.imshow(digits.images[i],cmap="gray")
    ax.set_title(f"Digit: {digits.target[i]}")
    ax.axis("off")

plt.show()
```

```python
#Write a program to count number of samples available in iris datsets and digit datasets
from sklearn.datasets import load_iris, load_digits

# Load datasets
iris = load_iris()
digits = load_digits()

# Count samples
iris_samples = len(iris.data)
digits_samples = len(digits.data)

# Display results
print("Number of samples in Iris dataset:", iris_samples)
print("Number of samples in Digits dataset:", digits_samples)
```
