```python
# ============================================================
# SEMANTIC NETWORK VISUALIZATION USING PYTHON
# ============================================================
#
# Required libraries:
# pip install networkx matplotlib
#
# ============================================================

import networkx as nx
import matplotlib.pyplot as plt


# ============================================================
# 1. CREATE GRAPH
# ============================================================

G = nx.Graph()


# ============================================================
# 2. DEFINE NODES
# ============================================================

nodes = [
    "Artificial Intelligence",
    "Machine Learning",
    "Deep Learning",
    "Neural Networks",
    "Natural Language Processing",
    "Computer Vision",
    "Robotics",
    "Data Science",
    "Python",
    "ChatGPT"
]

G.add_nodes_from(nodes)


# ============================================================
# 3. DEFINE SEMANTIC RELATIONSHIPS
# ============================================================

edges = [

    # AI relationships
    ("Artificial Intelligence", "Machine Learning"),
    ("Artificial Intelligence", "Natural Language Processing"),
    ("Artificial Intelligence", "Computer Vision"),
    ("Artificial Intelligence", "Robotics"),

    # Machine Learning relationships
    ("Machine Learning", "Deep Learning"),
    ("Machine Learning", "Data Science"),

    # Deep Learning relationships
    ("Deep Learning", "Neural Networks"),
    ("Deep Learning", "Computer Vision"),
    ("Deep Learning", "Natural Language Processing"),

    # NLP relationships
    ("Natural Language Processing", "ChatGPT"),
    ("Natural Language Processing", "Python"),

    # Computer Vision relationships
    ("Computer Vision", "Python"),

    # Data Science relationships
    ("Data Science", "Python"),

    # Robotics relationships
    ("Robotics", "Computer Vision"),
    ("Robotics", "Machine Learning"),

    # Neural Network relationship
    ("Neural Networks", "ChatGPT")
]

G.add_edges_from(edges)


# ============================================================
# 4. DEFINE NODE CATEGORIES
# ============================================================

categories = {

    "Artificial Intelligence": "AI",
    "Machine Learning": "AI",
    "Deep Learning": "AI",
    "Neural Networks": "AI",

    "Natural Language Processing": "Application",
    "Computer Vision": "Application",
    "Robotics": "Application",
    "ChatGPT": "Application",

    "Data Science": "Data",
    "Python": "Technology"
}


# ============================================================
# 5. DEFINE COLORS FOR CATEGORIES
# ============================================================

category_colors = {

    "AI": "skyblue",
    "Application": "lightgreen",
    "Data": "orange",
    "Technology": "pink"
}


# ============================================================
# 6. CREATE NODE COLORS
# ============================================================

node_colors = []

for node in G.nodes():

    category = categories[node]

    color = category_colors[category]

    node_colors.append(color)


# ============================================================
# 7. CALCULATE NODE SIZE BASED ON CONNECTIONS
# ============================================================

node_sizes = []

for node in G.nodes():

    degree = G.degree(node)

    # More connections = bigger node
    size = 1500 + (degree * 500)

    node_sizes.append(size)


# ============================================================
# 8. CREATE GRAPH LAYOUT
# ============================================================

pos = nx.spring_layout(
    G,
    k=1.5,
    iterations=100,
    seed=42
)


# ============================================================
# 9. CREATE FIGURE
# ============================================================

plt.figure(figsize=(18, 12))


# ============================================================
# 10. DRAW EDGES
# ============================================================

nx.draw_networkx_edges(
    G,
    pos,
    width=2,
    edge_color="gray",
    alpha=0.7
)


# ============================================================
# 11. DRAW NODES
# ============================================================

nx.draw_networkx_nodes(
    G,
    pos,
    node_color=node_colors,
    node_size=node_sizes,
    alpha=0.9,
    edgecolors="black",
    linewidths=1.5
)


# ============================================================
# 12. DRAW NODE LABELS
# ============================================================

nx.draw_networkx_labels(
    G,
    pos,
    font_size=10,
    font_weight="bold"
)


# ============================================================
# 13. CREATE LEGEND
# ============================================================

legend_items = []

for category, color in category_colors.items():

    item = plt.Line2D(
        [0],
        [0],
        marker="o",
        color="w",
        label=category,
        markerfacecolor=color,
        markeredgecolor="black",
        markersize=12
    )

    legend_items.append(item)


plt.legend(
    handles=legend_items,
    title="Categories",
    loc="upper left",
    fontsize=10,
    title_fontsize=12
)


# ============================================================
# 14. ADD TITLE
# ============================================================

plt.title(
    "Semantic Network of Artificial Intelligence Concepts",
    fontsize=20,
    fontweight="bold",
    pad=20
)


# ============================================================
# 15. ADD INFORMATION
# ============================================================

number_of_nodes = G.number_of_nodes()
number_of_edges = G.number_of_edges()

info_text = (
    f"Number of Concepts: {number_of_nodes}\n"
    f"Number of Relationships: {number_of_edges}"
)

plt.text(
    0.02,
    0.02,
    info_text,
    transform=plt.gca().transAxes,
    fontsize=11,
    bbox=dict(
        boxstyle="round",
        facecolor="white",
        edgecolor="black"
    )
)


# ============================================================
# 16. REMOVE AXIS
# ============================================================

plt.axis("off")


# ============================================================
# 17. ADJUST LAYOUT
# ============================================================

plt.tight_layout()


# ============================================================
# 18. SAVE GRAPH AS IMAGE
# ============================================================

plt.savefig(
    "semantic_network.png",
    dpi=300,
    bbox_inches="tight"
)


# ============================================================
# 19. DISPLAY GRAPH
# ============================================================

plt.show()


# ============================================================
# 20. PRINT NETWORK INFORMATION
# ============================================================

print("=" * 60)
print("SEMANTIC NETWORK INFORMATION")
print("=" * 60)

print("Number of Concepts:", G.number_of_nodes())
print("Number of Relationships:", G.number_of_edges())


print("\nCONCEPTS:")
print("-" * 60)

for node in G.nodes():

    print(
        f"{node} "
        f"(Category: {categories[node]}, "
        f"Connections: {G.degree(node)})"
    )


print("\nSEMANTIC RELATIONSHIPS:")
print("-" * 60)

for source, target in G.edges():

    print(
        f"{source}  <---->  {target}"
    )


print("\n" + "=" * 60)
print("Network visualization saved as: semantic_network.png")
print("=" * 60)
```
